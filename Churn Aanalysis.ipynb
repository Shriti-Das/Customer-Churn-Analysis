```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Customer-Churn.csv")
df
```

```python
df.info()
```

#Replacing blank space with 0 

```python
df["TotalCharges"]=df["TotalCharges"].replace(" ","0")
df["TotalCharges"]=df["TotalCharges"].astype("float")
```

```python
def conv(value):
    if(value==1):
        return "yes"
    else:
        return "no"
df["SeniorCitizen"]=df["SeniorCitizen"].apply(conv)
```

#Converted 0 and 1 to YES/NO for better understanding

```python
df.head(10)
```

```python
ax=sns.countplot(x="Churn",data=df)
ax.bar_label(ax.containers[0])
plt.title("No. of churned customers")
plt.show()
```

```python
plt.figure(figsize=(3,3))
gb=df.groupby("Churn").agg({"Churn":"count"})
plt.pie(gb['Churn'],labels=gb.index,autopct="%1.2f%%")
plt.title("Percentage of churned customers")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.countplot(x="gender",data=df,hue="Churn")
plt.title("Churn by Gender")
plt.show()
```

```python
plt.figure(figsize=(3,3))
sns.countplot(x="SeniorCitizen",data=df,hue="Churn")
plt.title("Churn by Senior Citizen")
plt.show()
```

```python
grouped = df.groupby(['SeniorCitizen', 'Churn']).size().unstack(fill_value=0)
percent = grouped.div(grouped.sum(axis=1), axis=0) * 100

# Step 2: Plotting
fig, ax = plt.subplots(figsize=(4, 4))

# Define colors (optional)
colors = ['#66c2a5', '#fc8d62']  # Customize as needed

# Plot stacked bars
bottom = [0] * len(percent)
for idx, churn_status in enumerate(percent.columns):
    bar = ax.bar(percent.index, percent[churn_status], bottom=bottom, label=churn_status, color=colors[idx])
    
    # Annotate each segment with percentage
    for i, val in enumerate(percent[churn_status]):
        if val > 0:
            ax.text(i, bottom[i] + val / 2, f'{val:.1f}%', ha='center', va='center', fontsize=8)
    bottom = [bottom[i] + percent[churn_status].iloc[i] for i in range(len(bottom))]

# Step 3: Final touches
ax.set_title('Percentage of Churn by Senior Citizen Status')
ax.set_ylabel('Percentage')
ax.set_xlabel('Senior Citizen')
ax.set_xticks([0, 1])
ax.set_xticklabels(['Not Senior', 'Senior'])
ax.legend(title='Churn')
plt.tight_layout()
plt.show()
```

#comparatively a greater number of senior citizen people have churned

```python
plt.figure(figsize=(5,4))
ax=sns.countplot(x="Contract",data=df,hue="Churn")
ax.bar_label(ax.containers[0])
plt.title("Churned customers by Contract")
plt.show()
```

#From above plot we can conclude that people having Month-to-month contract are more likely to churn than yearly contract people

```python
columns = ['PhoneService', 'MultipleLines', 'InternetService',
           'OnlineSecurity', 'OnlineBackup', 'DeviceProtection',
           'TechSupport', 'StreamingTV', 'StreamingMovies']

fig, axes = plt.subplots(3, 3, figsize=(15, 10))
axes = axes.flatten()

for i, col in enumerate(columns):
    sns.countplot(x=col, data=df, ax=axes[i],hue=df["Churn"])
    axes[i].set_title(col)
    axes[i].tick_params(axis='x', rotation=0)

plt.tight_layout()
plt.show()
```

#Customers without PhoneService or InternetService show very low churn, while those with services show a higher churn proportion.
#Features dependent on internet (like OnlineSecurity, OnlineBackup, TechSupport, etc.) tend to show higher churn when the customer has the service as "No".
#In general, customers lacking support or protection features are more likely to churn.Streaming services (StreamingTV and StreamingMovies) also show a similar pattern, where lack of service correlates with increased churn.
